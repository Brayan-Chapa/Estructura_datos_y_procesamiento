```python
import pandas as pd
SEPARADOR = ("*" * 60) + "\n"

diccionario_notas = {"Programación":[87, 80, 80, 20, 100],
                     "Base de Datos":[71,100,70,100,100],
                     "Contabilidad":[99,57,57,100,100]}

notas_diccionario = pd.DataFrame(diccionario_notas)
notas_diccionario.index = ["Crescensio", "Domitila","Rutilio","Panfilo","Ludoviko"]
print(notas_diccionario)
```

```python
print("\nLas notas obtenidas en programacion fueron: ")

print(notas_diccionario["Programación"])
print()

print(notas_diccionario.Programación)
print(SEPARADOR)
```

```python
print("Las notas obtenidas por Rutilio fueron: ")
print(notas_diccionario.loc["Rutilio"])
print("\nLas notas obtenidas por Ludoviko fueron: ")
print(notas_diccionario.loc["Ludoviko"])
print(SEPARADOR)
```

```python
print(f"\nLa menor calificación de Crescencio fue de: {notas_diccionario.loc['Crescensio'].min()}")
print()
print(f"El promedio de calificaciones de Crescencio fue de: {notas_diccionario.loc['Crescensio'].mean()}")
print(SEPARADOR)
```

```python
medias = notas_diccionario.mean()
print(type(medias))
print("\nPromedios por UA")
print("Se obtuvo la media de las siguientes asignaturas: ")

for materia in medias.index:
  print(materia)

print(f"\nEl promedio de las notas obtenidas fue de")
print(medias)
```

```python
media_global = notas_diccionario.mean().mean()
print(f"\nLa media global del DataFrame es de {media_global:.4f}")
```

```python
print("Las primeras dos materias para todos los estudiantes")
print(notas_diccionario[["Programación","Base de Datos"]])

print("\nLos ultimos dos estudiantes")
print(notas_diccionario.loc["Panfilo":"Ludoviko"])

print("\nLos estudiantes hasta Rutilio")
print(notas_diccionario.loc[:"Rutilio"])
```

```python
print("\nÚltimas dos asignaturas para todos los estudiantes a partir de Rutilio")
print(notas_diccionario[["Base de Datos","Contabilidad"]].loc["Rutilio":])

print(type(notas_diccionario[["Base de Datos","Contabilidad"]].loc["Rutilio":]))
print(SEPARADOR)
```

```python
print("La calificacion de Panfilo en Programacion fue de: ")
print(notas_diccionario.at["Panfilo","Programación"])
```

```python
filtro = notas_diccionario > 79

print("Este seria el filtro o mapa de bits resultante de aplicar la condicion")
print(filtro)
```

```python
print(notas_diccionario[notas_diccionario < 70])
print(SEPARADOR)

print(notas_diccionario[notas_diccionario < 70].min())
```

```python
base_datos_70_71 = notas_diccionario[notas_diccionario["Base de Datos"].isin([70,71])]

print(base_datos_70_71)
```

```python
filtro_aprobados_menor_100 = (notas_diccionario["Programación"] >= 70) & (notas_diccionario["Programación"] < 100)
aprobaron_progra_menos_que_100 = notas_diccionario.loc[filtro_aprobados_menor_100]

print(aprobaron_progra_menos_que_100)
print(SEPARADOR)
print(aprobaron_progra_menos_que_100["Programación"])
```
